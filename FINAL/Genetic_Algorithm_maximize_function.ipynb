Step 1: Initialize Population

```python
import random
def initPopulation(b,n): # n is the number of solutions, b is the length
    p = [random.choices([0,1],k = b) for i in range(n)]
    return p
```

```python
initPopulation(5,4)
```

Step 2: Reproduction or Parent Selection

Fitness Calculation~

```python
def binToDecimal(binList):
    decValue = 0 # decimal value
    power = len(binList)
    for digit in binList:
        decValue += digit*2**(power-1)
        power -= 1
    return decValue

def func(x):
    return x**2 # pow(x,2)

def getFitness(p): # p is the whole population
    f = [] # to store fitness values
    for sol in p:
        d = binToDecimal(sol)
        print('generated x : ',d)
        v = func(d)
        if v< 0: f.append(0)
        else: f.append(v) # fitness function
    return f
```

```python
populaion = initPopulation(5,4) # length 5, number of selection 4
getFitness(populaion)
```

```python
fitness = getFitness(populaion)
total_fitness = sum(fitness)
prob = [f/total_fitness for f in fitness]
n = len(populaion)
expected_count = [n*pi for pi in prob]
print(fitness)
print(prob)
print(expected_count)
```
