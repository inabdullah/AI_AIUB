Basic and Advance Arithmetic Operation

```python
#addition, sub, multip, div
a=10
b=5
sum= a+b
sub= a-b
prod= a*b
div= a/b
print("Sum is",sum)
print(f"Sub :{sub}, Product :{prod}, Division :{div}")
```

```python
# mod, pow, floor div
import math as m
c=2
d=3
mod=d%c
pow=m.pow(c,d) #2^3
pow2= c**d #c^d
fdiv=d//c #floor or int division
print(f"Mod:{mod}, power_fun: {pow}, Power_oper: {pow2}, and floor div: {fdiv}")
```

```python
# Absolute and Round
num1= -7.6
num2= 3.6
print("abs: ", abs(num1))
print("round num1:",round(num1))
print("round num2:",round(num2))
```

```python
#sqrt, factorial, trigonometric value
sqrt=int(m.sqrt(16))
fact=m.factorial(5)
sin_val=m.sin(m.radians(30))
print(f"Sqrt: {sqrt}, Factorial: {fact}, Sin30: {sin_val}")
```
