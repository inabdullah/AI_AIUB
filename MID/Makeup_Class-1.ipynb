```python
## function with multiple returns
def calc(a,b):
    sum=a+b
    sub=a-b
    mul=a*b
    return sum,sub,mul
```

```python
add, subtrac, prod= calc(3,4)
print(f"Add :{add}, Sub :{subtrac}, Mul :{prod}")
```
